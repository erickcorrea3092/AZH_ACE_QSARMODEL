```python
from rdkit.Chem import AllChem
from rdkit import Chem
from rdkit.Chem import Descriptors
from rdkit.ML.Descriptors import MoleculeDescriptors
import pandas as pd
import numpy as np
from mordred import Calculator, descriptors
```

```python
dataset = pd.read_excel('AZH_QSAR.xlsx')
dataset.shape
```

```python
dataset.tail()
```

```python
def canonical_smiles (smiles):
  mols = [Chem.MolFromSmiles(smi) for smi in smiles]
  smiles = [Chem.MolToSmiles(mol) for mol in mols]
  return smiles
```

```python
Canon_SMILES = canonical_smiles(dataset.SMILES)
len(Canon_SMILES)
```

```python
#put the smiles in dataframe
dataset['SMILES'] = Canon_SMILES
dataset
```

```python
#create a list for duplicate SMILES

duplicates_smiles = dataset[dataset['SMILES'].duplicated()]['SMILES'].values
len(duplicates_smiles)
```

```python
#Create a list for duplicate smiles
dataset[dataset['SMILES'].isin(duplicates_smiles)].sort_values(by=['SMILES'])
```

```python
dataset_new.reset_index(drop=True, inplace=True)
```

```python
def All_Mordred_descriptors(data):
  calc = Calculator(descriptors, ignore_3D=True)
  mols = [Chem.MolFromSmiles(smi) for smi in data]

  #pandas df
  df = calc.pandas(mols)
  return df
```

```python
mordred_descriptors = All_Mordred_descriptors(dataset_new['SMILES'])
```

```python
# prompt: combina dataset_new y mordred_descriptors y elimna la primera columna

# Combinar los dos DataFrames
combined_df = pd.concat([dataset_new, mordred_descriptors], axis=1)

# Eliminar la primera columna (asumiendo que es la primera columna de `dataset_new`)
# Si la primera columna de `mordred_descriptors` también necesita ser eliminada,
# se puede usar `combined_df = combined_df.iloc[:, 1:]`
combined_df = combined_df.iloc[:, 1:]

combined_df.tail()
```
